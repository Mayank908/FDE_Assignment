```python
####1.What is Dictionary in Python? List out all properties of dictionary .
# Dictionary is the collection of key-value pairs . Each Key is used to access its corresponding value.
# Example :- 

student = {
    "name" : "Mayank",
    "age" : 29,
    "course" : "FDE"
}

print(student["name"])

### Properties of Dictionary 
#1. Key-Value Pairs :- Data is stored in the form of key: value.
#2. Keys are unique :- Dictionary cannot have duplicate keys.
#3. Keys must be mutable :- Keys can be strings , numbers , tuples etc
#4. Values can be of any data type :- Values can be integers , strings , lists, dictionaries, etc
#5 Accessed using keys :- Values can be access by using thier kyes rather than indexes.
#6. Written using curly braces {} :- Dictionary elements are enclosed in curly brackets.
#7. Can be nested :- A dictionary can contain another dicionary as its value.
#8. Mutable :- A dictionary can be add, modify , or deleted items after creating a dictionary.
```

```python
####2. Explain the concept of the Dictionary Comprehension in Python.
# Dictionary comprehension is way to create dictionary in Python using single line of code.
#Syntax:
###{ key : value for item in iterable }

squares = { x: x*x  for x in range(1,6)}
print(squares)

# Here x is the key .
# x*x is the value
# range(1,6) provides the numbers from 1 to 5
```

```python
####3.Difference between dict.key(key) and dict[key]
#1. Using dict[key] => It key exists , it returns values otherwise it raises a KeyError.

student = {"name": "Rahul", "age": 20}
print(student['name'])

#print(student['marks'])

#2. Using dict.get(key) => It the key exists it returns , if not exists it returns None instead of producing and error .
student = {"name": "Rahul", "age":20}

print(student.get("name"))
print(student.get("marks"))

### Difference between both :- 
#####                          | `dict[key]`                              | `dict.get(key)`                   |
#####                          | ---------------------------------------- | --------------------------------- |
#####                          | Raises `KeyError` if key is missing      | Returns `None` if key is missing  |
#####                          | Cannot specify a default value           | Can specify a default value       |
#####                          | Useful when the key is expected to exist | Useful when the key may not exist |


```

```python
####4. How can you merge two dictionaries in python?

#1. Using update() method
dict1 = {"a" : 1 , "b" : 2}
dict2 = {"c" : 3 , "d" : 4}
dict1.update(dict2)
print(dict1)

#2. Using the | operator

dict1 = {"a" : 1 , "b" : 2}
dict2 = {"c" : 3 , "d" : 4}
merge = dict1 | dict2
print(merge)

#3. Using dictionary unpacking

dict1 = {"a" : 1 , "b" : 2}
dict2 = {"c" : 3 , "d" : 4}
merge = {**dict1, **dict2}
print(merge)
```

```python
####5. How can you update dictionary values in Python? Provide Examples
# Example : Update a single vlaue
student = {
    "name" : "Mayank",
    "age"  : 25,
    "marks": 75
}

student["marks"] = 85
print(student)

# Example : Update multiple values using update()
student = {
    "name" : "Mayank",
    "age"  : 25,
    "marks": 75
}

student.update({
    "age"   : 29,
    "marks" : 90
})

print(student)

# Example : Update a value a variable
student = {
    "name" : "Mayank",
    "age"  : 25,
    "marks": 75
}
new_marks = 85
student["marks"] = new_marks
print(student)
```

```python
####6. How can you iterate over a dictionary to access keys , values , and key-value pairs ?
#1. Iterate over keys
student = {
    "name" : "Mayank",
    "age"  : 25,
    "marks": 75
}
for key in student:
    print(key)
    
#2. Iterate over values
for value in student.values():
    print(value)

#3. Iterate over key-value pairs
for key , value in student.items():
    print(key, ":", value)
```

```python
####8. Count the frequency of each word in a given sentence using a dictionary.
sentence = "python is easy and python is powerful"
words = sentence.split()
frequency = {}

for word in words:
    if word in frequency:
        frequency[word] +=1
    else:
        frequency[word] = 1

print(frequency)
```

```python
####10. Sort a Dictionary by Keys and Values

#1. Sort Dictionary by Keys

student = {
    "banana" : 20,
    "apple"  : 30,
    "orange" : 10
}

sorted_by_keys = dict(sorted(student.items()))
print(sorted_by_keys)

#2. Sort Dictionary by values
student = {
    "banana" : 20,
    "apple"  : 30,
    "orange" : 10,
}

sorted_by_values = dict(sorted(student.items(), key=lambda item: item[1]))
print(sorted_by_values)
```

```python
#####13. Group a list of words into lists of anagrams using a dictionary
words = [
    "listen", "silent", "enlist",
    "hello",
    "lemon", "melon",
    "debitcard", "badcredit"
]
anagrams = {}

for word in words:
    key = ''.join(sorted(word))

    if key not in anagrams:
        anagrams[key] = []

    anagrams[key].append(word)
result = list(anagrams.values())

print(result)


```

```python
####14. Merge two dictionaries . If there are common keys , sum thier values.
dict1 = { "a" : 10 , "b": 20 , "c" : 30}
dict2 = { "b" : 5, "c" : 10 , "d" : 40}

merged = dict1.copy()

for key , value in dict2.items():
    if key in merged:
        merged[key] += value
    else:
        merged[key] = value

print(merged)
        


```

```python
####16. Find mirror characters in a string using a dictionary
text = "Hello World"
mirror = {}

for i in range(26):
    mirror[chr(65+i)] = chr(90-i)
    mirror[chr(97+i)] = chr(122-i)

result = ""

for char in text:
    if char in mirror:
        result += mirror[char]
    else:
        result += char

print(result)
    
```

```python
####22. Explain the difference between a set and frozenset
#Both set and frozenset store unique elements , but the difference is about mutablity 
# A set is mutable which means elements can be added or removed after creating it .
my_set = {1,2,3}
my_set.add(4)
my_set.remove(2)
print(my_set)

#A frozenset is immutable , which means we cannot add or remove elements after creating it 

my_frozenset = frozenset([1,2,3])
print(my_frozenset)
```

```python
####25. Find the minimum and maximum elements in a set without using min() and max().
numbers = {10,5,25,8,15}
first = True
for num in numbers:
    if first:
        minimum = num
        maximum = num
        first = False
    else:
        if num < minimum:
            minimum = num
        if num > maximum:
            maximum = num
print("Maximum" , maximum)
print("Minimum", minimum)
```

```python
####27. Find the intersection of two sets with and without using built-in methods
set1 = {1,2,3}
set2 = {3,4,5,6}
# Using built-in function
result = set1.intersection(set2)
print(result)
# Using the & operator
result = set1 & set2
print(result)
# Without using built-in function
set1 = {2,3,4,5,6}
set2 = {4,5,6,2}

result = set()

for item in set1:
    if item in set2:
        result.add(item)
print(result)
```

```python
####27. Find the symmetric differences of the two sets and without using built-in methods
# Symmetric difference refers to the elements are present in either set , but not in both sets.

set1 = {1,2,3,4}
set2 = {3,4,5,6}
#Using built in functions
result = set1.symmetric_difference(set2)
print(result)
#Using ^ operator
result = set1 ^ set2
print(result)
#Without using function
set1 = {1,2,3,4}
set2 = {3,4,5,6}

result = set()

for item in set1:
    if item not in set2:
        result.add(item)
for item in set2:
    if item not in set1:
        result.add(item)
print(result)

```

```python
####29. Perform set operations - union , intersection and difference - on sets containing both integers and strings , without using builtin functions

set1 = { 1, 2, "apple", "orange"}
set2 = { 2, 3, "orange", "banana"}
         
#Union
union = set1.copy()
for item in set2:
    union.add(item)
print("Union :", union)

#Intersection
intersection = set()
for item in set1:
    if item in set2:
        intersection.add(item)
print("Intersection :", intersection)

#Difference
difference = set()
for item in set1:
    if item not in set2:
        difference.add(item)
print("Difference :" , difference)


```
